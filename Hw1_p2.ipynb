```python
import numpy as np 
```

```python
array = np.zeros(10, dtype=int)
print(f"array({array})")
```

```python
array = np.ones(10, dtype=int) * 5
print(f"array({array})")
```

```python

array = np.arange(10, 50 + 1, 1, dtype=int)
print(f"array({array})")
```

```python
array = np.arange(10, 50 + 1, 2, dtype=int)
print(f"array({array})")
```

```python
def f1(i, j):
    return 3 * i + j
```

```python
matrix = np.fromfunction(f1, (3, 3), dtype=int)
print(f"matrix({matrix})")
```

```python
matrix = np.eye(3, dtype=int)
print(f"matrix({matrix})")
```

```python
random = np.random.random()
print(f"random({random})")
```

```python
random_array = np.random.rand(25)
print(f"random_array({random_array})")
```

```python
matrix = np.linspace(0.1, 1, 100).reshape(10, 10)
print(f"matrix({matrix})")
```

```python
mat = np.arange(1,26).reshape(5,5)

print(mat)

print(mat[2::, 1::])

print(mat[-3::, -4::])

print(mat[3, 4])

print(mat[(3, 4)])

print(mat[:3:, 1])

print(mat[:-2:, -4])

print(mat[-1])

print(mat[4])

print(mat[3::])

print(mat[-2::])

print(np.sum(mat))

print(np.median(mat))

print(np.sum(mat, axis=0))
```
